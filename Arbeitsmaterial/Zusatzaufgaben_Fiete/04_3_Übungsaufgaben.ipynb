**1) Berechnen Sie einen gewichteten Mittelwert der Zahlen im folgenden Dictionary. Werte, bei denen der Schlüssel mit einem "!" endet, sollen doppelt gewichtet werden. Vergleichen Sie mit dem ungewichteten Mittelwert.**

````python
messwerte = {'2018-10-19 20:58:17' : 29.52,
             '2018-10-19 23:31:09' : 41.72,
             '2018-10-19 18:27:14' : 13.68,
             '2018-10-20 00:23:12' : 27.39,
             '2018-10-19 21:24:47!': 21.92,
             '2018-10-20 01:37:01' : 34.70,
             '2018-10-19 19:43:20!': 20.85,
             '2018-10-19 20:31:55!': 29.27,
             '2018-10-19 23:57:24' : 27.28,
             '2018-10-19 19:17:05!': 47.96,
             '2018-10-19 23:06:13' : 18.15,
             '2018-10-19 18:01:29' : 30.15,
             '2018-10-20 00:48:26!': 20.16,
             '2018-10-19 22:39:59!': 25.25,
             '2018-10-19 21:49:38' : 27.05,
             '2018-10-19 20:07:21' : 39.39,
             '2018-10-19 18:52:01' : 32.54,
             '2018-10-19 17:37:20' : 20.06,
             '2018-10-19 22:14:07!': 33.50,
             '2018-10-20 01:13:07!': 41.34}


# Hier eigenen Code schreiben ...

mittel_ungewichtet = sum(messwerte.values())/len(messwerte)
mittel_gewichtet = 0
counter = 0
for i, j in messwerte.items():
    if '!' in i:
        mittel_gewichtet += 2*j
        counter += 2
    else:
        mittel_gewichtet += j
        counter += 1
mittel_gewichtet = mittel_gewichtet/counter

print(f'ungewichtet:{mittel_ungewichtet}\n gewichtet:{mittel_gewichtet}')
````

**2) Geben Sie das obige Dictionary als zeitlich sortierte Tabelle *ohne* "!" aus.**

```
 |  2018-10-19 17:37:20  |  20.06  |
 |  2018-10-19 18:01:29  |  30.15  |
 ...
```

````python
# Remove '!' from keys
for n in list(messwerte.keys()):
    if '!' in n:
        messwerte[n[:-1]] = messwerte.pop(n)

# Sort by timestamp and print as table
print("Zeitlich sortierte Messungen:\n")
for timestamp in sorted(messwerte.keys()):
    print(f" |  {timestamp}  |  {messwerte[timestamp]}  |")
````

**3) Geben Sie die Messzeiten aus, zu denen die `n` geringsten Werte gemessen wurden.** 

Hinweis: Je nach Lösungsansatz kann es hilfreich sein, die Daten selbst während der Verarbeitung zu verändern.
Generell sollte man in solchen Fällen darauf achten, dass die Ursprungsdaten dennoch erhalten bleiben. 
Das ist recht einfach möglich, indem nur eine Kopie der Daten verwendet wird. 
Diese kann z.B. so erstellt werden:
```python
messkopie = messwerte.copy()
``` 

Es ist überdies auch möglich, nicht (mehr) benötigte Variablen wieder aus dem Speicher zu entfernen:
```python
del mein_variablenname
``` 
Dies ist z.B. dann sinnvoll, wenn während des Skriptdurchführung Speicher freigegeben werden muss oder bestimmte Daten nicht mehr in die weitere Verarbeitung einfließen sollen.

````python
n = 3

print(f"Die {n} geringsten Messwerte:\n")
for timestamp, value in sorted(messwerte.items(), key=lambda x: x[1])[:n]:
    print(f"{timestamp}: {value}")
````

**4) Wieviel [Zeit](https://docs.python.org/3/library/time.html) (in Minuten) vergeht jeweils zwischen zwei aufeinanderfolgenden Messungen?**

````python
import time

# Sort timestamps
sorted_times = sorted(messwerte.keys())

print("Zeitdifferenzen zwischen aufeinanderfolgenden Messungen:\n")

# Loop through pairs of consecutive timestamps
for i in range(len(sorted_times) - 1):
    time1_str = sorted_times[i]
    time2_str = sorted_times[i + 1]
    
    # Parse timestamp strings to time objects
    time1 = time.strptime(time1_str, '%Y-%m-%d %H:%M:%S')
    time2 = time.strptime(time2_str, '%Y-%m-%d %H:%M:%S')
    
    # Convert to seconds
    seconds1 = time.mktime(time1)
    seconds2 = time.mktime(time2)
    
    # Calculate difference in minutes
    minutes = (seconds2 - seconds1) / 60
    
    print(f"{time1_str} → {time2_str}: {minutes:.1f} Minuten")
````
