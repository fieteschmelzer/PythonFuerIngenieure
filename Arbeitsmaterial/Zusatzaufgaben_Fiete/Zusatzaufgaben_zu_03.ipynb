**1.<br>Schreiben Sie eine Funktion `pol2cart()`, die einen Punkt aus 2-dimensionalen [Polarkoordinaten](https://de.wikipedia.org/wiki/Polarkoordinaten) (in radians) in kartesische Koordinaten umrechnet!**
* Beispiel: `pol2cart(1, 2)` &rarr;  `-0.416, 0.909`
* Plotten Sie anschließend den Punkt mit den Polarkoordinaten 1, 2 mit `plot()` und dem Parameter `'*'`!

```python
import numpy as np
import math
import matplotlib.pyplot as plt
from math import pi

def pol2cart(betrag, winkel):
    '''
    Rechne die 2-dimensionalen Polarkoordinaten eines Punktes in
    kartesische Koordinaten um!
    
    Parameters:
    -----------
    betrag: int or float
        Der Abstand des Punktes vom Koordinatenursprung.
    winkel: int or float
        Der Winkel zwischen der Polachse (x-Achse) und dem vom
        Koordinatenursprung durch den Punkt führenden Strahl.
    
    Returns:
    -----------
    x: float
        Der Abstand zwischen dem Koordinatenursprung und der
        Projektion des Punktes auf die x-Achse.
    y: float
        Der Abstand zwischen dem Koordinatenursprung und der
        Projektion des Punktes auf die y-Achse.
    ''' 
    # Diesen sogenannten Docstring haben wir schon im ersten Notebook
    # "00_Jupyter.ipynb" kennengelernt. Er lässt sich über ein
    # nachgestelltes Fragezeichen abrufen, also "pol2cart?"
    
    # Hier eigenen Code schreiben...

    x = np.cos(winkel)* betrag
    y = np.sin(winkel)* betrag
    return x, y

x, y = pol2cart(1, 2)

plt.plot(x, y, '*')

    
```

* Die Punkte sollen in Polarkoordinaten angegeben sein.
* Die Punkte sollen auf einer gemeinsamen Kreisbahn mit dem Koordinatenursprung als Mittelpunkt liegen.
* Der Radius des Kreises soll als optionaler Parameter angegeben werden können (Standardwert: 1).
* Der Abstand zwischen zwei nebeneinanderliegenden Punkten soll jeweils gleich sein.
* Der erste und der letzte Eck-Punkt sollen identisch sein (damit das Polygon beim Plotten vollständig gezeichnet wird).

**Zeichnen Sie ein Fünfeck mit Radius 2!**

```python
# Hier eigenen Code schreiben...

def polygon(Ecken, Radius = 1):
    winkel = 2*np.pi/Ecken
    winkel_lis = np.zeros(Ecken+1)
    dis = np.zeros(Ecken+1)
    dis[0] = Radius
    winkel_lis[0] = np.pi/2
    
    for i in range(0, Ecken):
        winkel_lis[i+1] = winkel_lis[i]+winkel
        dis[i+1] = Radius
    return winkel_lis, dis

win, distance = polygon(9, 2)

x, y = pol2cart_list(distance, win)



plt.plot(x, y, '*')
plt.axis('equal')
plt.show()
```

**6.<br>
[Ada](https://de.wikipedia.org/wiki/Ada_Lovelace) und [Grace](https://de.wikipedia.org/wiki/Grace_Hopper) fliegen nach Las Vegas und betrinken sich heftig an der Minibar.<br>
Welche der folgenden Handlungen hat am nächsten Tag, wenn sie wieder ausgenüchtert aus Las Vegas abreisen, bleibende Konsequenzen und welche "bleiben in Vegas"?**
* Ada möchte an ihrem Manuskript schreiben, nimmt **Feder** und **Tinte** und das **Manuskript** aus ihrem Koffer, verschüttet aber sofort die Tinte über ihr Manuskript und steckt **Feder** und **bekleckertes Manuskript** beschämt zurück in den Koffer.
* Ada und Grace verwechseln ihre Koffer.
* Ada öffnet "Ihren" Koffer, entfernt den **Asteroiden** und wirft ihn aus dem Fenster, weil sie keine Steine mag.
* Grace stellt erschüttert fest, dass sie ihre **Motte** und ihren **Asteroiden** in "ihrem" Koffer nicht finden kann, geht in die Stadt und kauft sich einen **neuen leeren Koffer**, der jetzt Ihr Koffer sein soll.
* Grace findet den **Asteroiden** vor dem Hotel, freut sich und gibt ihm nach ihrem Ehemann den Namen **Vincent**.
* Grace fängt sich eine **Motte** und steckt sie und **Vincent** in "ihren" Koffer.

**Wie erklären Sie sich, welche Handlungen auch außerhalb der lokalen Funktion globale Folgen haben und welche nicht?**

```python
def betrunken_in_vegas(adas_koffer, graces_koffer):
    # Ada: Tinte verschüttet, Manuskript bekleckert (modifiziert Liste - PERSISTS global!)
    adas_koffer.remove('Tinte')
    adas_koffer.remove('Manuskript')
    adas_koffer.append('bekleckertes Manuskript')
    
    # Koffer verwechseln (nur lokale Umbenennung - NO global effect!)
    adas_koffer, graces_koffer = graces_koffer, adas_koffer
    
    # Ada wirft Asteroid raus aus "ihrem" Koffer (ist jetzt original graces_koffer)
    # Das modifiziert weiterhin die Originalliste!
    if 'Asteroid' in adas_koffer:
        adas_koffer.remove('Asteroid')
    
    # Grace kauft neuen Koffer (lokale Variable - NO global effect!)
    neuer_koffer = []
    graces_koffer = neuer_koffer
    
    # Grace findet Asteroid und nennt ihn Vincent (lokale Variable - NO effect!)
    vincent = 'Asteroid (jetzt Vincent)'
    
    # Grace steckt Motte und Vincent in ihren neuen lokalen Koffer
    graces_koffer.append('Motte')
    graces_koffer.append(vincent)
    
    print(f'In Vegas:\n "Adas Koffer": {adas_koffer}\n "Graces Koffer": {graces_koffer}\n Vincent: {vincent}\n')
```

```python
adas_koffer = ['Buch', 'Manuskript', 'Feder', 'Tinte']
graces_koffer = ['Motte', 'Asteroid']
vincent = 'Graces Ehemann'

print(f'Vor Vegas:\n Adas Koffer: {adas_koffer}\n Graces Koffer: {graces_koffer}\n Vincent: {vincent}\n')

betrunken_in_vegas(adas_koffer, graces_koffer)

print(f'Nach Vagas:\n Adas Koffer: {adas_koffer}\n Graces Koffer: {graces_koffer}\n Vincent: {vincent}')
```

Grace und Ada trinken bei der Arbeit Tee. Sie gehen gemeinsam in die Küche und beschließen dort, als nächstes gemeinsam Kaffee zu trinken. Ada hat Ihre Tasse aber an ihrem Schreibtisch stehen lassen.**
* Grace gibt Ada ihre Tasse, nimmt sich eine neue leere Tasse und füllt Kaffee hinein.
* Ada trinkt den Tee aus ihrer Tasse und füllt Kaffee hinein.

**Implementieren Sie die Funktion!<br>
Geben Sie den Inhalt der Tassen vor, in und nach der Funktion mit `print()` aus!<br>
Wie erklären Sie sich die Ausgabe?**

```python
def in_der_küche(adas_tasse, graces_tasse):
    print(f'In der Küche (Anfang):\n Adas Tasse: {adas_tasse}\n Graces Tasse: {graces_tasse}\n')
    
    # Grace gibt Ada ihre Tasse (nur lokale Umbenennung - NO global effect!)
    adas_tasse = graces_tasse
    
    # Grace nimmt sich eine neue leere Tasse (lokale Variable - NO global effect!)
    graces_tasse = []
    
    # Grace füllt ihre neue Tasse mit Kaffee
    graces_tasse.append('Kaffee')
    
    # Ada trinkt den Tee aus ihrer Tasse (die lokal graces_tasse ist!) und füllt mit Kaffee
    adas_tasse.remove('Tee')
    adas_tasse.append('Kaffee')
    
    print(f'In der Küche (Ende):\n Adas Tasse: {adas_tasse}\n Graces Tasse: {graces_tasse}\n')
```

```python
adas_tasse = ['Tee']
graces_tasse = ['Tee']

print(f'Vor der Küche:\n Adas Tasse: {adas_tasse}\n Graces Tasse: {graces_tasse}\n')

in_der_küche(adas_tasse, graces_tasse)

print(f'Nach der Küche:\n Adas Tasse: {adas_tasse}\n Graces Tasse: {graces_tasse}')

# Erklärung: 
# - adas_tasse = graces_tasse ist nur eine lokale Umbenennung (NO effect)
# - graces_tasse = [] ist auch nur lokal (NO effect)
# - ABER: adas_tasse.remove() und .append() modifizieren das Objekt selbst (PERSISTENT!)
# - Das global graces_tasse war ursprünglich das gleiche Objekt wie lokal adas_tasse
# - Deswegen hat der Tee→Kaffee Wechsel in graces_tasse bleibende Folgen!
```
